# Step 7: Split Data into Training and Testing Sets
The dataset is split into training and testing sets for model evaluation.

```python
x = df.iloc[:,0:-1]
```

```python

y = df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

# Step 8: Train Decision Tree
We train a Decision Tree Classifier on the training data. You have to also check for KNeighborsClassifier and  GaussianNB

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
df = DecisionTreeClassifier()
```

```python
df.fit(x_train, y_train)
```

```python
y_pred = df.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print("accuracy : ", accuracy_score(y_test, y_pred))
```

```python
df.score(x_test, y_pred)
```

# Cross Validation

```python
from sklearn.model_selection import KFold
```

```python
k = KFold(n_splits=10)
```

```python
x
```

```python
for train_index,test_index in k.split(x):
    x_train, x_test = x.values[train_index],x.values[test_index]
    y_train, y_test = y[train_index],y[test_index]
    model1 = DecisionTreeClassifier()
    model1.fit(x_train,y_train)
    print(model1.score(x_test,y_test))

```

```python
from sklearn.model_selection import cross_val_score
```

```python
cross_val_score(df,x,y,cv=10)
```

# BS4 | web Scraping to get Data

```python
import requests
import bs4
```

```python
req = requests.get("https://darshan.ac.in/faculty-list/btech-computer")
```

```python
req.status_code
```

```python
soup = bs4.BeautifulSoup(req.text, "lxml")
```

```python
data = soup.select("body > main > div:nth-child(5) > div > div > div > div > div")
```

```python
len(data)
```

```python
for row in data:
    name = row.select("h2")[0].text.strip()
    other = row.select("span")
    print(name,"---->",other[0].text.strip())
```
