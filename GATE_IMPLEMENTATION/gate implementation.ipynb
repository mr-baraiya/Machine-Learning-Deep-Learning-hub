```python
import numpy as np
```

```python
# 1 if positive, o if negative
def step_function(x):
    if x >= 0:
        return 1
    else:
        return 0
print("Setup Complete.")
```

```python
#inputs
x_or = np.array([[0,0],[0,1],[1,0],[1,1]])
x_or
```

```python
#targets
# y_or =  np.array([0,0,0,1]) # AND Gate
y_or =  np.array([0,1,1,1]) # OR Gate
# y_or =  np.array([0,1,1,0]) # X-OR Gate
y_or
```

```python
weights =np.array([0,0])
bias = 0
lr = 0.5 # learning rate
print(f'Initrial Rate : {weights}')
print(f'Initrial Bias : {bias}')
```

```python
# Example Input : (0,1)
input_sample = x_or[1]
target_value = y_or[1]

#1. weighted sum (Dot Product)
z = np.dot(input_sample,weights) + bias
print(f"Math:({input_sample[0]} * {weights[0]}) + ({input_sample[1]} * {weights[1]}) + {bias}")
print(f"Sum (z) : {z:.2f}")

#2. Activation
prediction = step_function(z)
print(f"Prediction : {prediction}")

#3. Error
error = target_value - prediction
print(f'Error : {error}')
```

```python
epochs = 5
print("---TRAINING---")
for epoch in range(epochs):
    print(f"\nEPOCH-> {epoch+1}")

    for i in range(4):
        x = x_or[i]
        y = y_or[i]

        #predict
        z = np.dot(x,weights) + bias
        pred = step_function(z)

        # Error
        error = y - pred

        # Learn (Update Weights)
        if error !=0 :
            weights = weights + (lr * error * x)
            bias = bias + (lr * error * 1) # always one
            print(f'MISTAKE on {x}. Expected {y} ,got {pred}.')
            print(f'Updated weights: {weights}, Bias: {bias:.2f}.')
        else:
            print(f"Coreect On {x}.")

print("Final Weights : ",weights)
print("Final Bias : ",bias)
```
